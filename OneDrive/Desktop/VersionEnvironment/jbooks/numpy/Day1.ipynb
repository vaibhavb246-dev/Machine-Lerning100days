#Array Reshaping

```python
ar=np.arange(12)
print(ar)
reshaped=ar.reshape((3,4))
print(reshaped)
flatend=reshaped.flatten()
print(flatend)
```

```python
reveled=reshaped.ravel()
print(reveled)
```

#Transpose

```python
transpose=reshaped.T
print(transpose)
```
