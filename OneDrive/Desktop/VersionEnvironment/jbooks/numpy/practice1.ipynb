```python
import numpy as np
```

```python
a = np.array([1, 2, 3])                    # 1-D
b = np.array([[1, 2, 3], [4, 5, 6]])       # 2-D
c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
```

```python
print(np.linspace(0, 1, 5))  
print(np.logspace(0, 3, 4))   
print(np.empty((2, 2)))  
print(np.zeros_like(b), np.ones_like(b))    
print(np.full_like(b, 9)) 
x, y = np.meshgrid(np.arange(3), np.arange(2))
print(x, y, sep="\n")
print(np.fromfunction(lambda i, j: i * 10 + j, (3, 3), dtype=int))
print(np.tri(3), np.triu(np.ones((3, 3))), np.tril(np.ones((3, 3))), sep="\n")
```

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print("shape   :", a.shape)
print("ndim    :", a.ndim)
print("size    :", a.size)
print("dtype   :", a.dtype)
print("itemsize:", a.itemsize, "bytes")
print("nbytes  :", a.nbytes)
print("strides :", a.strides)
f = a.astype(np.float32)
print(f.dtype)
print(np.array([1, 2, 3], dtype=np.int8))
print(np.array([1, 2.5, 3]).dtype)          # upcasting -> float64
print(np.array(["a", "bb"]).dtype)          # <U2
print(np.array([True, False]).dtype)
print(np.array([1 + 2j]).dtype)             # complex128
print(np.iinfo(np.int32), np.finfo(np.float32).eps)
print(np.result_type(np.int8, np.float32))
# Overflow example
print(np.array([250], dtype=np.uint8) + np.uint8(10))
print(np.result_type(np.int8, np.float32))
print(np.iinfo(np.int32), np.finfo(np.float32).eps)
```

```python
a= np.arange(1,13).reshape(3,4)
print(a)
print(a[0,1],a[-1,-1])
```

```python
print(a[0],a[:,0])
print(a[0:2, 1:3])
print(a[::2, ::2])
print(a[:,None,:].shape)
```

```python
v = a[0:2, 0:2]
v[0, 0] = 999
print("original changed:", a[0, 0])
c = a.copy()                                # real copy
c[0, 0] = -1
print("original untouched:", a[0, 0])
```

```python
a = np.arange(12)
print(a.reshape(3, 4))
print(a.reshape(-1, 2)) 
m = a.reshape(3, 4)
print(m.T) 
print(m.ravel())
print(m.flatten())  

print(np.expand_dims(a, 0).shape, np.squeeze(np.zeros((1, 3, 1))).shape)
print(np.swapaxes(np.zeros((2, 3, 4)), 0, 2).shape)
print(np.moveaxis(np.zeros((2, 3, 4)), 0, -1).shape)
print(np.transpose(np.zeros((2, 3, 4)), (2, 0, 1)).shape)
print(np.roll(a, 3))
print(np.flip(m, axis=1))
print(np.rot90(m))
print(np.tile([1, 2], 3), np.repeat([1, 2], 3))
print(np.pad([1, 2, 3], (2, 2), mode="constant"))
print(np.resize(a, (2, 3)))
```
