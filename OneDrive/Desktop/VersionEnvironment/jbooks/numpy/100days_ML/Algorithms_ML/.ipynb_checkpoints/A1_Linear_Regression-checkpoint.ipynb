```python
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

# Generating random dataset

```python
np.random.seed(100)

X = np.random.rand(50,1)*100

Y = 3.5* X + np.random.randn(50,1)*20
```

# Creating and training linear regression model 

```python
model = LinearRegression()
model.fit(X , Y)
```

#  Preding Y values

```python
Y_pred = model.predict(X)
```

# Visualizing the Regression Line

```python
plt.figure(figsize=(9,6))
plt.scatter(X,Y,colorizer='blue',label='Data points')
plt.plot(X,Y_pred,color='red',linewidth =2,label ="regression line")
plt.title('Linear regression on random dataset')
plt.xlabel('hours')
plt.ylabel('marks')
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(9,6))
plt.scatter(X,Y,colorizer='blue',label='Data points')
plt.plot(X,Y,color='red',linewidth =2,label ="regression line")


plt.show()
```

# slope and  Intercept

```python
print("Slope (Coefficient):", model.coef_[0][0])
print("Intercept:",model.intercept_[0])
```

# 2.Price prediction of pizza using linear regression

```python
import pandas as pd 
```

```python
data = {
    'size':[6,8,10,12,14],
    'price':[100,150,200,250,300]
}

df =pd.DataFrame(data)
```

```python
df
```

```python
df.describe()
```

```python
plt.scatter(df['size'],df['price'])

plt.xlabel('Size of Pizza(inch)')
plt.ylabel('price of pizza (rupee)')
plt.title('Pizza price prediction')

plt.show()
```

```python
# Seprate input and target
X = df[['size']]
Y = df[['price']]

#create   model
model = LinearRegression()

# train model
model.fit(X,Y)

# display lear values 
print('Slope:',model.coef_[0])
print('Intercept:',model.intercept_)
```

```python
new_pizza = pd.DataFrame({'size':[15.5]})

pred = model.predict(new_pizza)

print('pizza size',15.5)
print('predict price:',pred[0])
```

```python
new = pd.DataFrame({'size':[7,9.2,13.6,14.1,21.2]})

new['Predicted_price'] = model.predict(new[['size']])

print(new)
```

# display graph

```python
plt.scatter(X['size'],Y,label='Actual price')
plt.plot(
    X['size'],
    model.predict(X),
    color = 'pink',
    linewidth=2,
    label = 'Regression line'
)
plt.xlabel('pizza size')
plt.ylabel('pizza price')

plt.title('Pizza price prediction using Linear regression')
plt.legend()
plt.show()
```
