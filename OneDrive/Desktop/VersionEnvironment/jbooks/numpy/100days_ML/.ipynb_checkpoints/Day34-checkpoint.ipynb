# Handling Date And Time Variable 

```python
import pandas as pd 
import numpy as np
```

```python
date = pd.read_csv('orders.csv')
time = pd.read_csv('messages.csv')
```

# Working with date

```python
date['date']= pd.to_datetime(date['date'])
```

```python
date.info()
```

# 1.Extract year

```python
# date['date']=pd.to_date(date['date'])
date['date_year']=date['date'].dt.year
date.head()
```

# 2.Extract Month

```python
date['date_month_no']= date['date'].dt.month
date.head()
```

```python
date['date_month_name']= date['date'].dt.month_name()

date.head()
```

# 4.Extract Days

```python
date['date_day']=date['date'].dt.day
date.head()
```

```python
date['date_dow'] = date['date'].dt.dayofweek

date.head()
```

```python
# day of week name 
date['date_dow_name']= date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# is weekend 
date['date_is_weekend']= np.where(date['date_dow_name'].isin(['Saturday','Sunday']),1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

# Extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week 

date.drop(columns=['product_id','city_id','orders']).head()
```

#  Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

# Extract Semester

```python
date['semester']= np.where(date['quarter'].isin([1,2]),1,2)

date.drop(columns=['product_id','city_id','orders']).head()
```

# Extract Time elapsed between dates

```python
import datetime
today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed 

# np.round((today - date['date'])/np.timedelta64(1,'M'),0)

# 30.44 days is the standard average month length across a leap-year cycle
# np.round((today - date['date']) / np.timedelta64(1, 'M') / 30.44, 0)

# Calculates the absolute grid difference in calendar months
(today.year - date['date'].dt.year) * 12 + (today.month - date['date'].dt.month)


```

```python
np.round((today - date['date']) / np.timedelta64(1, 'D') / 30.44, 0)
```
