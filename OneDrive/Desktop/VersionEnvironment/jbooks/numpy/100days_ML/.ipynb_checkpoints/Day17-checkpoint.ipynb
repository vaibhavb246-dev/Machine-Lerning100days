```python
import time
import requests
import pandas as pd

df_list = []

for i in range(1, 1429):
    url = f"https://api.themoviedb.org/3/movie/popular?api_key=65b9d4fd940870ab58f257a800f0cf81" # your full URL
    
    try:
        r = requests.get(url)
        
        # Only proceed if the request was successful (Status 200)
        if r.status_code == 200:
            data = r.json()
            if "results" in data:
                temp_df = pd.DataFrame(data['results'])[['id', 'title', 'overview', 'popularity', 'release_date', 'vote_average', 'vote_count']]
                df_list.append(temp_df)
        else:
            print(f"Page {i} failed with status code: {r.status_code}")
            
    except Exception as e:
        print(f"An error occurred on page {i}: {e}")
        
    # Crucial: Pause for 0.2 seconds between requests to avoid getting blocked
    time.sleep(0.2)

# Check if we actually gathered any data before concatenating
if df_list:
    df = pd.concat(df_list, ignore_index=True)
    print(f"Successfully created DataFrame with {len(df)} rows.")
else:
    print("DataFrame could not be created because no data was successfully fetched.")
```
