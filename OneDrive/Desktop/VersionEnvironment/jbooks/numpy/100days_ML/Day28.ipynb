```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv('covid_toy.csv')
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

# 1.normaly

```python
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[['fever']])

X_test_fever= si.fit_transform(X_test[['fever']])
X_train_fever.shape
```

```python
oe=OrdinalEncoder(categories= [['Mild','Strong']])
X_train_cough=oe.fit_transform(X_train[['cough']])

X_test_cough= oe.fit_transform(X_test[['cough']])
```

```python
X_train_cough.shape
```

```python
ohe=OneHotEncoder(drop='first',)
X_train_gender=ohe.fit_transform(X_train[['gender','city']]).toarray()

X_test_gender= ohe.fit_transform(X_test[['gender','city']]).toarray()
```

```python
X_train_gender.shape
```

```python
X_train_age=X_train.drop(columns=['gender','fever','cough','city']).values

X_test_age=X_test.drop(columns=['gender','fever','cough','city']).values
```

```python
X_train_age.shape
```

```python
X_train_transformed= np.concatenate((X_train_age,X_train_fever,X_train_gender,X_train_cough),axis=1)

X_test_transformed= np.concatenate((X_test_age,X_test_fever,X_test_gender,X_test_cough),axis=1)
```

# 2. By using Column Transform

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer(transformers=[('a1',SimpleImputer(),['fever']),
                                            ('a2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
                                            ('a3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
                                           ],remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
